# Assignment_1: Data Pre Processing using Functions in Python:

```python
import pandas as pd
import numpy as np
import seaborn as sns

df = pd.read_csv("ipl_data.csv")
df.head()
```

```python
df.isnull().sum()
```

# for NUMERICAL VALUES -  replace by mean

```python
def heart_killers(data):
    num_cols = data.select_dtypes(include=['int64', 'float64']).columns

    for col in num_cols:
        data[col] = data[col].fillna(data[col].mean())

    return data
```

# for CATEGORICAL VALUES - replace by mode

```python
def double_helix(data):
    cat_cols = data.select_dtypes(include=['object', 'string']).columns

    for col in cat_cols:
        data[col] = data[col].fillna(data[col].mode()[0])

    return data
```

```python
df = heart_killers(df)
df = double_helix(df)
```
